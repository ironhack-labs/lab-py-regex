### Exercise 1: Match Email Addresses
Write a regex pattern to match valid email addresses. An email address should contain an '@' symbol and a domain.

```python
import re

# Example input
email = "test@example.com"

# Your regex pattern here
pattern = r"\w+@\w+\.\w+"  # matches basic email format: one or more word characters before '@', domain and top-level domain

# Test the regex
if re.match(pattern, email):
    print("Valid email")
else:
    print("Invalid email")
```

### Exercise 2: Validate Phone Numbers
Create a regex pattern to validate phone numbers in the format (123) 456-7890 or 123-456-7890.

```python
import re

# Example input
phone_number = "(123) 456-7890"

# Your regex pattern here
pattern = r"\(\d{3}\) \d{3}-\d{4}"
# matches phone numbers in the format (123) 456-7890

# Test the regex
if re.match(pattern, phone_number):
    print("Valid phone number")
else:
    print("Invalid phone number")
```

### Exercise 4: Match URLs
Create a regex pattern to match URLs that start with http:// or https://.

```python
import re

# Example input
url = "https://www.example.com"

# Your regex pattern here
pattern = r"https?://\S+"  # matches URLs starting with http:// or https:// followed by non-whitespace characters

# Test the regex
if re.match(pattern, url):
    print("Valid URL")
else:
    print("Invalid URL")
```

### Exercise 5: Find Words Starting with a Specific Letter
Write a regex pattern to find all words starting with the letter 'a' in a given string.

```python
import re

# Example input
color_code = "#FFFFFF"

# Your regex pattern here
pattern = r"^#(?:[0-9A-Fa-f]{6})$"

          # ^ → Start of the string
          # # → Matches the literal hash symbol
          # (?: ... ) → Non-capturing group
          # [0-9A-Fa-f]{6} → Matches 6 hexadecimal digits
          # $ → End of the string

# Test the regex
if re.match(pattern, color_code):
    print("Valid hex color code")
else:
    print("Invalid hex color code")
```

### Exercise 7: Validate Passwords
Write a regex pattern to validate passwords that must be at least 8 characters long and contain at least one uppercase letter, one lowercase letter, one digit, and one special character.

```python
import re

# Example input
password = "Password123!"

# Your regex pattern here
pattern = r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$"

          # ^ → Start of the string
          # (?=.*[a-z]) → Must contain at least one lowercase letter
          # (?=.*[A-Z]) → Must contain at least one uppercase letter
          # (?=.*\d) → Must contain at least one digit
          # (?=.*[@$!%*?&]) → Must contain at least one special character
          # [A-Za-z\d@$!%*?&]{8,} → Must be at least 8 characters long and only contain allowed characters
          # $ → End of the string

# Test the regex
if re.match(pattern, password):
    print("Valid password")
else:
    print("Invalid password")
```

### Exercise 8: Remove Extra Spaces
Create a regex pattern that removes extra spaces from a string while keeping single spaces between words.

```python
import re

# Example input
text = "This   is   an   example."

# Your regex pattern here
pattern = r"\s{2,}"

          # \s → steht für whitespace (Leerzeichen, Tabs, Zeilenumbrüche usw.)
          # {2,} → bedeutet „zwei oder mehr Wiederholungen“
          # re.sub(pattern, " ", text) → ersetzt alle doppelten/mehrfachen Leerzeichen durch genau ein Leerzeichen

# Replace extra spaces
cleaned_text = re.sub(pattern, " ", text)
print(cleaned_text)
```

### Exercise 9: Match IP Addresses
Write a regex pattern to match valid IPv4 addresses.

```python
import re

# Example input
ip_address = "192.168.1.1"

# Your regex pattern here
pattern = r"^(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)(?:\.(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)){3}$"

          # ^ → start of the string
          # (?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d) → matches one valid octet (0 – 255)
          # 25[0-5] → 250 – 255
          # 2[0-4]\d → 200 – 249
          # 1\d{2} → 100 – 199
          # [1-9]?\d → 0 – 99
          # (?:\.(...)){3} → repeats the same octet pattern three more times, each preceded by a dot
          # $ → end of the string

# Test the regex
if re.match(pattern, ip_address):
    print("Valid IP address")
else:
    print("Invalid IP address")
```

### Bonus Exercise 2: Validate Credit Card Numbers  
Create a regex pattern to validate credit card numbers (16 digits).

```python
import re

# Example input
credit_card_number = "1234-5678-9876-5432"

# Your regex pattern here
pattern = r"^(?:\d{4}[- ]?){3}\d{4}$"

          # ^ → start of the string
          # (?:\d{4}[- ]?){3} → three groups of 4 digits with optional dash or space
          # \d{4} → final group of 4 digits
          # $ → end of the string

# Test the regex
if re.match(pattern, credit_card_number):
    print("Valid credit card number")
else:
    print("Invalid credit card number")
```

### Bonus Exercise 3: Match Non-Alphanumeric Characters  
Write a regex pattern to match non-alphanumeric characters in a string.

```python
import re

# Example input
text = "Hello! How are you? @Python3."

# Your regex pattern here
pattern = r"[^A-Za-z0-9\s]"

          # [^A-Za-z0-9\s] -> matches any non-alphanumeric character (excluding whitespace)

# Find all matches
non_alphanumeric_chars = re.findall(pattern, text)
print(non_alphanumeric_chars)
```

### Bonus Exercise 4: Validate Date Format  
Create a regex pattern to validate dates in the format DD/MM/YYYY.

```python
import re

# Example input
date_string = "25/12/2024"

# Your regex pattern here
pattern = r"\d{2}/\d{2}/\d{4}"

          # \d{2}/\d{2}/\d{4} → matches dates in DD/MM/YYYY format

# Test the regex
if re.match(pattern, date_string):
    print("Valid date format")
else:
    print("Invalid date format")
```

### Bonus Exercise 5: Extract Email Domains  
Write a regex pattern to extract domains from email addresses.

```python
import re

# Example input
email_list = ["user@example.com", "admin@domain.org"]

# Your regex pattern here
pattern = r"\w+@\w+\.\w+"

for email in email_list:
    domain = re.search(pattern, email)
    if domain:
        print(domain)  # Print extracted domain part.
```
